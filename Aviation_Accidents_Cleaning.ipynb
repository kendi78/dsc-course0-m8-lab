### Cleaning and constructing Key Measurables

Injuries and robustness to destruction are a key interest point for the client. Clean and impute relevant columns and then create derived fields that best quantifies what the client wishes to track. **Use commenting or markdown to explain any cleaning assumptions as well as any derived columns you create.**

**Construct metric for fatal/serious injuries**

*Hint:* Estimate the total number of passengers on each flight. The likelihood of serious / fatal injury can be estimated as a fraction from this.

```python
# --- Load dataset ---
aviation = pd.read_csv("data/AviationData.csv", encoding="latin-1")
```

```python
# Step 1: Clean injury columns
# Convert injury columns to numeric and fill NaNs with 0
injury_cols = ['Total.Fatal.Injuries','Total.Serious.Injuries',
               'Total.Minor.Injuries','Total.Uninjured']

for col in injury_cols:
    aviation[col] = pd.to_numeric(aviation[col], errors='coerce').fillna(0)
print("✅ Injury columns cleaned:")
print(aviation[injury_cols].head())
```

Assumption: Missing values mean no injuries were reported, so we impute with 0

```python
#Step 2
aviation['Total.Occupants'] = aviation[injury_cols].sum(axis=1)
print("\n✅ Total occupants (first 5 rows):")
print(aviation[['Total.Occupants']].head())
```

**Derived field**: Total.Occupants approximates the number of passengers  and crew.

```python
# step 3
aviation['Severe.Injury.Fraction'] = (
    (aviation['Total.Fatal.Injuries'] + aviation['Total.Serious.Injuries']) /
    aviation['Total.Occupants'].replace(0, pd.NA)
)

print("\n✅ Severe injury fraction (first 5 rows):")
print(aviation[['Severe.Injury.Fraction']].head())
```

**Derived field**: Severe.Injury.Fraction quantifies the likelihood of serious/fatal injury per accident.
**Assumption**: If Total.Occupants = 0, we mark as NaN to avoid division errors.

**Aircraft.Damage**
- identify and execute any cleaning tasks
- construct a derived column tracking whether an aircraft was destroyed or not.

```python
aviation['Aircraft.damage'] = aviation['Aircraft.damage'].astype(str).str.title().fillna("Unknown")

aviation['Destroyed.Flag'] = aviation['Aircraft.damage'].apply(
    lambda x: 1 if x == "Destroyed" else 0
)

print("\n✅ Aircraft damage categories (unique values):")
print(aviation['Aircraft.damage'].unique())

print("\n✅ Destroyed flag counts:")
print(aviation['Destroyed.Flag'].value_counts())
```

**Derived field**: Destroyed.Flag is a binary indicator of whether the aircraft was destroyed.
**Assumption**: "Unknown" or NaN are treated as not destroyed (conservative bias).

### Investigate the *Make* column
- Identify cleaning tasks here
- List cleaning tasks clearly in markdown
- Execute the cleaning tasks
- For your analysis, keep Makes with a reasonable number (you can put the threshold at 50 though lower could work as well)

**Standardize case**

Convert all entries to title case (e.g., "BOEING" → "Boeing").

**Handle missing values**

Replace NaN with "Unknown".

**Remove obvious junk entries**

Drop rows where Make is "Unknown" or meaningless codes.

**Filter by frequency**

Keep only manufacturers with at least 50 records (adjustable threshold).

```python
# Step 1: Standardize case and fill missing
aviation['Make'] = aviation['Make'].astype(str).str.title().replace("Nan", "Unknown")

# Step 2: Check unique values before filtering
print("\n✅ Unique Makes (sample):")
print(aviation['Make'].unique()[:20])  # show first 20 unique values
```

```python
# Step 3: Count frequencies
make_counts = aviation['Make'].value_counts()
print("\n✅ Top 10 Makes by frequency:")
print(make_counts.head(10))
```

```python
# Step 4: Apply threshold filter (>= 50 records)
valid_makes = make_counts[make_counts >= 50].index
aviation = aviation[aviation['Make'].isin(valid_makes)]
print("\n✅ Remaining Makes after filtering (>=50 records):")
print(aviation['Make'].value_counts())
```

### Column Removal
- inspect the dataframe and drop any columns that have too many NaNs

```python
# Count missing values per column
na_counts = aviation.isna().sum()

# Total rows in dataset
total_rows = len(aviation)

# Calculate % of NaNs per column
na_percent = (na_counts / total_rows) * 100

print("✅ Percentage of NaNs per column:")
print(na_percent.sort_values(ascending=False))
```

**> 40 NANs sparse drop**

```python
# Define threshold for dropping (e.g., >40% NaNs)
threshold = 40  

# Calculate NaN percentages again
na_counts = aviation.isna().sum()
na_percent = (na_counts / len(aviation)) * 100

# Identify columns to drop
cols_to_drop = na_percent[na_percent > threshold].index

print("✅ Columns to drop (too many NaNs):")
print(cols_to_drop)

# Drop them
aviation = aviation.drop(columns=cols_to_drop)

print("\n✅ Remaining columns after drop:")
print(aviation.columns)
```
